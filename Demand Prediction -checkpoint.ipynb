# Data Visualization

## 1. Pie Chart of Customer distribution acroos Distribution Centers

```python
DCcount = df['DC'].value_counts()
DCcount
```

```python
explode = (0.03 , 0 , 0)
plt.figure(figsize = (8,6))
plt.pie(DCcount , labels = DCcount.index, autopct='%1.1f%%', startangle=140 , explode = explode, colors= ["#00ebc1", "#00b8ba", "#0085b4"],data = df)
plt.title('Customer Distribution Acroos Distribution Centers')
plt.show
```

## 2. Inventory Level per Distribution Center

```python
plt.figure(figsize = (10 , 6))
ax = sns.barplot(df , x = 'DC', y= 'Adjusted Inventory', palette= 'pastel' , legend = False)
plt.title('Inventory Level by Distribution Centers through Disruption')

for container in ax.containers:
    ax.bar_label(container, label_type='edge', fmt='%d')
plt.show    
```

## 3. Box plot for Customer Demand Distribution

```python
df_melted = df.melt(id_vars='Customers', value_vars=['Min', 'Max', 'Mean'],
                    var_name='Demand_Type', value_name='Demand')
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_melted, x='Demand_Type', y='Demand', palette = 'pastel')
plt.title('Customer Demand Distribution by Min, Max, and Mean')
plt.xlabel('Demand Type')
plt.ylabel('Demand (Units)')
plt.show()
```

## 4. Stacked Bar Chart for Transit Times across Different Routes

```python
df_grouped = df.groupby('DC').mean(numeric_only = True)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(df_grouped.index, df_grouped['Supplier to HK Port Time'], label='Supplier to HK')
plt.bar(df_grouped.index, df_grouped['HK Port to Marseilles Port Time'], bottom=df_grouped['Supplier to HK Port Time'], label='HK to Marseilles')
plt.bar(df_grouped.index, df_grouped['Marseilles Port to DC Time'], bottom=df_grouped['Supplier to HK Port Time'] + df_grouped['HK Port to Marseilles Port Time'], label='Marseilles to DC')
plt.bar(df_grouped.index, df_grouped['DC to Customer Time'], bottom=df_grouped['Supplier to HK Port Time'] + df_grouped['HK Port to Marseilles Port Time'] + df_grouped['Marseilles Port to DC Time'], label='DC to Customer')

plt.xlabel('Route')
plt.ylabel('Transit Time (Days)')
plt.title('Transit Times across Different Routes')
plt.legend(loc='upper left')
plt.show()
```

## 5. Heat map for Demand Distribution per Region

```python
region_demand = df.groupby('Region')['Adjusted Mean'].sum().reset_index()
region_demand
```

```python
plt.figure(figsize = (10 , 6))
sns.heatmap(region_demand.set_index('Region'), annot=True, cmap='coolwarm' , fmt = 'g')
plt.title("Heatmap of Demand by Region")
plt.xlabel("Demand")
plt.ylabel("Region")
plt.show()
```

# Machine Learning Model

```python
x = df[['Min','Max','Mean','Lockdown Delays','Total Transit Time','Supplier to HK Port Time','HK Port to Marseilles Port Time','Marseilles Port to DC Time','DC to Customer Time','Inventory','Adjusted Inventory','DC_encoded','Region_encoded']].values
y = df[['Adjusted Mean']].values
print(x[0:5])
print(y[0:5])
```

```python
from sklearn import preprocessing
```

```python
Scaler = preprocessing.StandardScaler().fit(x)
x = Scaler.transform(x)
x[0:5]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
RFR = RandomForestRegressor(n_estimators = 100 , random_state = 42)
RFR.fit(x_train , y_train)

y_prediction = RFR.predict(x_test)
```

```python
MAE = mean_absolute_error(y_test , y_prediction)
print('Mean Absolute Error is:' , MAE)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_prediction)
print("R² Score:", r2)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_prediction)
rmse = np.sqrt(mse)

print("MSE:", mse)
print("RMSE:", rmse)
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(RFR , x , y , cv = 5)
print("Cross-validation scores:", scores)
print("Mean cross-validation score:", scores.mean())
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
knn = KNeighborsRegressor(n_neighbors = 2)
knn.fit(x_train , y_train)

y_hat = knn.predict(x_test)
```

```python
mae = mean_absolute_error(y_test, y_hat)
MSE = mean_squared_error(y_test, y_hat)
rMSE = np.sqrt(mse)
r2score = r2_score(y_test, y_hat)

print(f'MAE: {mae}, MSE: {mse}, R^2: {r2} ,RMSE: {rMSE}')
```
